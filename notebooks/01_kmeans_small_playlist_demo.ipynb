```python
# 1. import the model
from sklearn.cluster import KMeans
import pandas as pd
import matplotlib.pyplot as plt
from adjustText import adjust_text
```

```python
# import csv file
music_df = pd.read_csv('../data/spotify_10_songs.csv')
```

```python
music_df
```

```python


# strip whitespace from column names
music_df = music_df.set_index('song_name')
```

```python

# Drop the 'artist', 'id', and 'html' columns

music_df = music_df.drop(columns=['artist', 'id', 'html'])

```

```python
music_df
```

```python
# 2. initialize the model

random_state = 42

my_music_kmeans = KMeans(n_clusters= 4,
                         random_state=random_state)

# fit te model to the data
my_music_kmeans.fit(music_df)

# assign labels to the data
table = my_music_kmeans.labels_

# attach the labels to the original dataframe
music_df['cluster'] = table

music_df.sort_values(by='cluster')
```

```python
# calculate the mean of the danceability and show the 3 songs
danceability_songs = music_df['danceability'].sort_values(ascending=False)
danceability_songs
```

```python
energy_songs = music_df['energy'].sort_values(ascending=False)
energy_songs
```

```python
spechness_songs = music_df['speechiness'].sort_values(ascending=False)
spechness_songs
```

```python
accousticness_songs = music_df['acousticness'].sort_values(ascending=False)
accousticness_songs
```

```python
# create copy of music_df
music_df = music_df.copy()
```

```python
# create a df for the columns danceability and energy
dance_energy_df = music_df.iloc[:, :2].copy()
dance_energy_df
```

```python
dance_energy_kmeans = KMeans(n_clusters= 4,
                         random_state=random_state)

# fit te model to the data
dance_energy_kmeans.fit(dance_energy_df)

# assign labels to the data
table = dance_energy_kmeans.labels_

# attach the labels to the dataframe
dance_energy_df['cluster'] = table

dance_energy_df.sort_values(by='cluster')
```

```python
# Scatter plot of clusters - Danceability vs Energy
plt.figure(figsize=(10, 6))

for cluster in sorted(dance_energy_df['cluster'].unique()):
    cluster_data = dance_energy_df[dance_energy_df['cluster'] == cluster]
    
    plt.scatter(
        cluster_data['danceability'],
        cluster_data['energy'],
        label=f'Cluster {cluster}',
        alpha=0.7,
        s=100
    )

# Labels and title
plt.xlabel('Danceability', fontsize=12)
plt.ylabel('Energy', fontsize=12)
plt.title('Cluster Distribution: Danceability vs Energy', fontsize=14)

# Legend
plt.legend(fontsize=10)

# Grid
plt.grid(True, alpha=0.3)

# Show plot
plt.tight_layout()
plt.show()
```

```python

# Cluster label mapping
cluster_labels = {
    0: 'Chill / Bossa Nova',
    1: 'Energetic Groove',
    2: 'High-Energy Rap',
    3: 'Classical / Soft Acoustic'
}

# Create figure
plt.figure(figsize=(10, 6))

# Plot clusters
for cluster in dance_energy_df['cluster'].unique():
    
    cluster_data = dance_energy_df[dance_energy_df['cluster'] == cluster]
    
    plt.scatter(
        cluster_data['danceability'],
        cluster_data['energy'],
        s=120,
        alpha=0.8,
        label=cluster_labels[cluster]
    )

# Song labels
for i, row in dance_energy_df.iterrows():
    
    plt.text(
        row['danceability'] + 0.01,
        row['energy'] + 0.01,
        i,
        fontsize=8
    )

# Labels and title
plt.xlabel('Danceability')
plt.ylabel('Energy')
plt.title('Music Clusters Based on Danceability and Energy')

# Grid and legend
plt.grid(True)
plt.legend(title='Music Type')

# Show plot
plt.show()
```

```python
# create df with energy and acousticness
energy_accousticness_df = music_df.iloc[:,[1,6]].copy()
energy_accousticness_df
```

```python
energy_accousticness_kmeans = KMeans(n_clusters= 4,
                         random_state=random_state)

# fit the model to the data
energy_accousticness_kmeans.fit(energy_accousticness_df)

# assign labels to the data
table = energy_accousticness_kmeans.labels_

# attach the labels to the dataframe
energy_accousticness_df['cluster'] = table

energy_accousticness_df.sort_values(by='cluster')
```

```python

cluster_labels = {
    0: 'Chill Bossa Nova',
    1: 'Electronic / Rock / Rap',
    2: 'Balanced Groove',
    3: 'Classical / Acoustic'
}

plt.figure(figsize=(12, 7))

texts = []

for cluster in energy_accousticness_df['cluster'].unique():
    cluster_data = energy_accousticness_df[energy_accousticness_df['cluster'] == cluster]
    
    plt.scatter(
        cluster_data['energy'],
        cluster_data['acousticness'],
        s=120,
        alpha=0.8,
        label=cluster_labels[cluster]
    )

for song_name, row in energy_accousticness_df.iterrows():
    texts.append(
        plt.text(
            row['energy'],
            row['acousticness'],
            song_name,
            fontsize=8
        )
    )

adjust_text(
    texts,
    arrowprops=dict(arrowstyle='-', lw=0.5)
)

plt.xlabel('Energy')
plt.ylabel('Acousticness')
plt.title('Music Clusters: Energy vs Acousticness')
plt.grid(True)
plt.legend(title='Music Type')
plt.show()
```

```python


cluster_labels = {
    0: 'Chill Bossa Nova',
    1: 'Electronic / Rock / Rap',
    2: 'Balanced Groove',
    3: 'Classical / Acoustic'
}

plt.figure(figsize=(11, 7))

for cluster in energy_accousticness_df['cluster'].unique():
    cluster_data = energy_accousticness_df[energy_accousticness_df['cluster'] == cluster]
    
    plt.scatter(
        cluster_data['energy'],
        cluster_data['acousticness'],
        s=140,
        alpha=0.8,
        label=cluster_labels[cluster]
    )

# Add short numbers instead of full song names
for number, (song_name, row) in enumerate(energy_accousticness_df.iterrows(), start=1):
    plt.text(
        row['energy'] + 0.01,
        row['acousticness'] + 0.01,
        str(number),
        fontsize=10,
        weight='bold'
    )

plt.xlabel('Energy')
plt.ylabel('Acousticness')
plt.title('Music Clusters: Energy vs Acousticness')
plt.grid(True)
plt.legend(title='Music Type')
plt.show()

# Show number-to-song mapping
for number, song_name in enumerate(energy_accousticness_df.index, start=1):
    print(f"{number}: {song_name}")
```
